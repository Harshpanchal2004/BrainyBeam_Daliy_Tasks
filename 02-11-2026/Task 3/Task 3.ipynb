```python
import os
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", font_scale=1.05)
plt.rcParams['font.sans-serif'] = 'Arial'
```

```python
data_path = "womens_clothing_classification_dataset.csv"
if not os.path.exists(data_path):
    data_path = os.path.join("02-11-2026", "Task 3", "womens_clothing_classification_dataset.csv")

df = pd.read_csv(data_path)
df['Full_Review_Text'] = df['Full_Review_Text'].fillna('')
print(f"Loaded Women's Clothing Classification Dataset: {df.shape[0]:,} reviews.")
```

```python
pages_csv = "application_page_routing_specifications.csv"
if not os.path.exists(pages_csv):
    pages_csv = os.path.join("02-11-2026", "Task 3", "application_page_routing_specifications.csv")

df_pages = pd.read_csv(pages_csv)
display(df_pages)
```

```python
fig, ax = plt.subplots(figsize=(11, 5.5))
colors = ['#1A365D', '#2B6CB0', '#2E8540', '#319795', '#D69E2E', '#805AD5']
bars = ax.barh(df_pages['Page_Title'][::-1], df_pages['Average_Render_Time_ms'][::-1], color=colors[::-1], alpha=0.88, edgecolor='black', linewidth=0.8)
ax.axvline(25.0, color='#C53030', linestyle='--', linewidth=1.5, label='Maximum Render Target (25 ms)')
ax.set_title("Streamlit View Mount & Rendering Latency Benchmarks", fontsize=12, fontweight='bold')
ax.set_xlabel("Render Latency (ms)", fontsize=10.5)
ax.legend(loc='lower right', frameon=True)

for bar in bars:
    w = bar.get_width()
    ax.text(w + 0.3, bar.get_y() + bar.get_height()/2.0, f"{w:.1f} ms", va='center', fontsize=9.5, fontweight='bold')

plt.tight_layout()
plt.show()
```
